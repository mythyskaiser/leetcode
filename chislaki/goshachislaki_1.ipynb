```python
import math
import numpy as np
import matplotlib.pyplot as plt

k,p,D,A = 1,1,1,1
l = 3
# h = 0.1
# x_arr = np.arange(0,1,h)
# print(x_arr)
# print("  fgff")
def float_range(start,end,step):
    while start<end:
        yield start
        start += step

def f(x):
    res = k*(1-((x*x+p)/D)*x)-((x*x+p)/D)*A
    return res
def dyldx(x,y):
    dyldx = ((x*x+p)/D)*y + f(x)
    return dyldx
y0 = A
x0 = 0
dots_x = []
dots_y = []
errors = []
def right_f(x):
    res = k*x + A
    return res
def approximator(h):
    y_i = y0
    x_i = x0
    for i in float_range(0,l,h):
        y_i_1 = y_i + h*dyldx(x_i,y_i)
        dots_x.append(x_i)
        dots_y.append(y_i_1)
        error_i = y_i_1 - right_f(x_i)
        errors.append(error_i)
        y_i = y_i_1
        x_i += h
approximator(0.01)
plt.plot(dots_x, dots_y, marker = 'o', linestyle ='--', color = 'red')
print(errors)
plt.grid()
plt.show()
```

```python
import math
import numpy as np
import matplotlib.pyplot as plt

k,p,D,A, q = 1,1,1,1,1
l = 4
# h = 0.1
# x_arr = np.arange(0,1,h)
# print(x_arr)
# print("  fgff")
def float_range(start,end,step):
    while start<end:
        yield start
        start += step

def f(x):
    res = -5*math.exp(x)*(1+p-q)
    return res
def fi(x,u,z):
    res = q*u-p*z-f(x)
    return res
z0 = 50
u0 = 0
x0 = 0
dots_x = []
dots_y = []
errors = []
def right_f(x):
    res = 5*np.exp(x)
    return res
def approximator(h):
    z_i = z0
    u_i = u0
    x_i = x0
    for i in float_range(0,2.4,h):
        k1 = h*z_i
        l1 = h*fi(x_i,u_i,z_i)
        k2 = h*(z_i+l1/2)
        l2 = h*fi(x_i+h/2,u_i+k1/2, z_i+l1/2)
        k3 = h*(z_i+l2/2)
        l3 = h*fi(x_i+h/2, u_i+k2/2, z_i+l2/2)
        k4 = h*(z_i+l3)
        l4 = h*fi(x_i+h,u_i+k3,z_i+l3)
        delta_u0 = 1/6*(k1+2*k2+2*k3+k4)
        delta_z0= 1/6*(l1+2*l2+2*l3+l4)
        error_i = abs( right_f(x_i)-fi(x_i,u_i,z_i))
        u_i = u_i+delta_u0
        z_i = u_i+delta_z0
        x_i += h
        dots_x.append(x_i)
        dots_y.append(u_i)
        print(fi(x_i,u_i,z_i), "chisl")
        print(right_f(x_i), "given")
        errors.append(error_i)
approximator(0.11)
plt.plot(dots_x, dots_y, marker = 'o', linestyle ='--', color = 'red')
x = np.linspace(0, 2.4, 20) 

# 2. Define the exponential function
# Calculate y = e^x
y = 5*np.exp(x)
plt.plot(x,y, marker= 'o', color='blue' )
print(errors)
plt.grid()
plt.show()
print()
```

```python
import math
import matplotlib.pyplot as plt
import numpy as np
y0 = np.array([1,0,0])
x0 = 0
x1 = 10
h = 0.00001

def dydx(y):
    y1, y2, y3 = y
    f1 = -0.04 * y1 + 1e4 * y2 * y3
    f2 = 0.04 * y1 - 1e4 * y2 * y3 - 3e7 * (y2**2)
    f3 = 3e7 * (y2**2)
    return np.array([f1, f2, f3])

n = (1-0)/h
y_arr = []
f_arr = []
def approximator(n: int):
    k = 0
    y_n = y0
    while k<3:
        k+=1
        k1 = dydx(y_n)
        k2 = dydx(y_n+h/2*k1)
        k3 = dydx(y_n+h/2*k2)
        k4 = dydx(y_n+h*k3)
        f_arr.append(dydx(y_n))
        y_arr.append(y_n)
        y_n = y_n + h/6*(k1 + 2*k2 + 2*k3 + k4)
    div = 0
    k-=1
    while k<n:
        y_n = y_arr[k] + h/12*(23*f_arr[k]-16*f_arr[k-1]+5*f_arr[k-2])
        div = dydx(y_n)
        y_n = y_arr[k] + h/12*(5*div+8*f_arr[k]-f_arr[k-1])
        y_arr.append(y_n)
        f_arr.append(dydx(y_n))
        k+=1

approximator(n)
plt.figure(figsize=(10, 5))
plt.plot(np.linspace(0,10, len(y_arr)), y_arr)
plt.show()
print(y_arr[-1])
```
