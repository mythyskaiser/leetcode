```python
import numpy as np
import matplotlib.pyplot as plt
import math

# Вариант а)
a = 2.0
a2 = a**2
L = math.pi
T = 1.0

def f_fun(x, t):
    return np.cos(x / 2) * np.exp(t)

def u_exact(x, t):
    return math.pi - x + 0.5 * (np.exp(t) - np.exp(-t)) * np.cos(x / 2)

def apply_left_bc(y, h):
    # u_x(0,t) = -1
    # (-3u0 + 4u1 - u2) / (2h) = -1
    y[0] = (4*y[1] - y[2] + 2*h) / 3
    y[-1] = 0
    return y

def progonka(a_low, b_diag, c_up, d):
    # a_low[0] не используется, c_up[-1] не используется
    n = len(d)
    p = np.zeros(n)
    q = np.zeros(n)
    x = np.zeros(n)

    p[0] = -c_up[0] / b_diag[0]
    q[0] = d[0] / b_diag[0]

    for i in range(1, n):
        den = b_diag[i] + a_low[i] * p[i-1]
        p[i] = -c_up[i] / den if i < n-1 else 0
        q[i] = (d[i] - a_low[i] * q[i-1]) / den

    x[-1] = q[-1]
    for i in range(n-2, -1, -1):
        x[i] = p[i] * x[i+1] + q[i]

    return x

def make_grid(N):
    h = L / N

    # Для явной схемы нужно r = a^2*tau/h^2 <= 1/2
    tau0 = 0.45 * h**2 / a2

    M = math.ceil(T / tau0)
    tau = T / M

    x = np.linspace(0, L, N + 1)
    t = np.linspace(0, T, M + 1)

    return x, t, h, tau, M

def start_layer(x, h):
    y = math.pi - x
    y[-1] = 0
    y = apply_left_bc(y, h)
    return y

def explicit_scheme(N):
    x, t, h, tau, M = make_grid(N)
    r = a2 * tau / h**2

    y = np.zeros((M + 1, N + 1))
    y[0] = start_layer(x, h)

    for j in range(M):
        y[j+1, 1:N] = (
            y[j, 1:N]
            + r * (y[j, 0:N-1] - 2*y[j, 1:N] + y[j, 2:N+1])
            + tau * f_fun(x[1:N], t[j])
        )

        y[j+1, N] = 0
        y[j+1] = apply_left_bc(y[j+1], h)

    return x, t, y, h, tau, r

def implicit_scheme(N):
    x, t, h, tau, M = make_grid(N)
    r = a2 * tau / h**2
    n = N - 1

    y = np.zeros((M + 1, N + 1))
    y[0] = start_layer(x, h)

    for j in range(M):
        rhs = y[j, 1:N] + tau * f_fun(x[1:N], t[j+1])

        a_low = np.zeros(n)
        b_diag = np.ones(n) * (1 + 2*r)
        c_up = np.zeros(n)

        a_low[1:] = -r
        c_up[:-1] = -r

        # Левая граница: y0=(4y1-y2+2h)/3
        b_diag[0] = 1 + 2*r/3
        c_up[0] = -2*r/3
        rhs[0] += 2*r*h/3

        y[j+1, 1:N] = progonka(a_low, b_diag, c_up, rhs)

        y[j+1, N] = 0
        y[j+1] = apply_left_bc(y[j+1], h)

    return x, t, y, h, tau, r

def crank_nicolson_scheme(N):
    x, t, h, tau, M = make_grid(N)
    r = a2 * tau / h**2
    n = N - 1

    y = np.zeros((M + 1, N + 1))
    y[0] = start_layer(x, h)

    for j in range(M):
        tm = t[j] + tau/2

        rhs = (
            y[j, 1:N]
            + 0.5*r*(y[j, 0:N-1] - 2*y[j, 1:N] + y[j, 2:N+1])
            + tau * f_fun(x[1:N], tm)
        )

        a_low = np.zeros(n)
        b_diag = np.ones(n) * (1 + r)
        c_up = np.zeros(n)

        a_low[1:] = -0.5*r
        c_up[:-1] = -0.5*r

        # Левая граница: y0=(4y1-y2+2h)/3
        b_diag[0] = 1 + r/3
        c_up[0] = -r/3
        rhs[0] += r*h/3

        y[j+1, 1:N] = progonka(a_low, b_diag, c_up, rhs)

        y[j+1, N] = 0
        y[j+1] = apply_left_bc(y[j+1], h)

    return x, t, y, h, tau, r

def max_error(x, y, time):
    return np.max(np.abs(y - u_exact(x, time)))

# Основной расчет
N = 50

schemes = [
    ("Явная", explicit_scheme),
    ("Неявная", implicit_scheme),
    ("Симметричная", crank_nicolson_scheme)
]

print("Вариант а)")
print("Точное решение:")
print("u(x,t)=pi-x+0.5*(e^t-e^(-t))*cos(x/2)")
print()

results = {}

for name, scheme in schemes:
    x, t, y, h, tau, r = scheme(N)
    err = max_error(x, y[-1], T)
    results[name] = (x, t, y, h, tau, r, err)

    print(name)
    print("h =", h)
    print("tau =", tau)
    print("r =", r)
    print("max error at T =", err)
    print()

# График решения в момент T
plt.figure(figsize=(9, 5))

xx = results["Явная"][0]
plt.plot(xx, u_exact(xx, T), label="Точное", linewidth=2)

for name in results:
    x, t, y, h, tau, r, err = results[name]
    plt.plot(x, y[-1], "--", label=name)

plt.xlabel("x")
plt.ylabel("u(x,T)")
plt.title("Сравнение решений при T=1")
plt.grid(True)
plt.legend()
plt.show()

# График ошибки
plt.figure(figsize=(9, 5))

for name in results:
    x, t, y, h, tau, r, err = results[name]
    plt.plot(x, np.abs(y[-1] - u_exact(x, T)), label=name)

plt.xlabel("x")
plt.ylabel("|ошибка|")
plt.title("Ошибка численного решения при T=1")
plt.grid(True)
plt.legend()
plt.show()

# Таблица зависимости ошибки от шага
print("Зависимость ошибки от N")
print("N        M        explicit        implicit        symmetric")

for Ntest in [20, 40, 80]:
    x1, t1, y1, h1, tau1, r1 = explicit_scheme(Ntest)
    x2, t2, y2, h2, tau2, r2 = implicit_scheme(Ntest)
    x3, t3, y3, h3, tau3, r3 = crank_nicolson_scheme(Ntest)

    e1 = max_error(x1, y1[-1], T)
    e2 = max_error(x2, y2[-1], T)
    e3 = max_error(x3, y3[-1], T)

    print(f"{Ntest:<8d} {len(t1)-1:<8d} {e1:<15.6e} {e2:<15.6e} {e3:<15.6e}")
```

```python
import numpy as np
from math import exp

# -----------------------------
# Параметры задачи
# -----------------------------
a = 2.0
l1 = 1.0
l2 = 1.0

N1 = 20
N2 = 20

h1 = l1 / N1
h2 = l2 / N2

eps = 1e-6
max_iter = 300000


# -----------------------------
# Точное решение
# -----------------------------
def u_exact(x, y):
    return a * exp(x + y)


# -----------------------------
# Коэффициенты варианта 1
# -----------------------------
def k1(x, y):
    return x**2 + y**2 + 10


def k2(x, y):
    return 5 * (x**2 + y**2) + 20


def p1(x, y):
    return x + y**2 + 5


def p2(x, y):
    return y + x**2 + 10


def q(x, y):
    return 2*x + 10*y + x**2 + y**2 + 10


def f_fun(x, y):
    # Взято так, чтобы точное решение u = 2e^(x+y) удовлетворяло уравнению
    return -(6*(x**2 + y**2) + x + y + 35) * u_exact(x, y)


# -----------------------------
# Нумерация внутренних узлов
# -----------------------------
def number(i, j):
    return (j - 1) * (N1 - 1) + (i - 1)


# -----------------------------
# Построение матрицы системы M y = b
# -----------------------------
def build_system():
    n = (N1 - 1) * (N2 - 1)

    M = np.zeros((n, n))
    b = np.zeros(n)

    for j in range(1, N2):
        y = j * h2

        for i in range(1, N1):
            x = i * h1
            row = number(i, j)

            k_e = k1(x + h1 / 2, y)
            k_w = k1(x - h1 / 2, y)

            k_n = k2(x, y + h2 / 2)
            k_s = k2(x, y - h2 / 2)

            A_e = k_e / h1**2 + p1(x, y) / (2 * h1)
            A_w = k_w / h1**2 - p1(x, y) / (2 * h1)

            A_n = k_n / h2**2 + p2(x, y) / (2 * h2)
            A_s = k_s / h2**2 - p2(x, y) / (2 * h2)

            A_c = - (k_e + k_w) / h1**2 - (k_n + k_s) / h2**2 - q(x, y)

            M[row, row] = A_c

            rhs = -f_fun(x, y)

            # Правая точка
            if i + 1 <= N1 - 1:
                M[row, number(i + 1, j)] = A_e
            else:
                rhs -= A_e * u_exact(l1, y)

            # Левая точка
            if i - 1 >= 1:
                M[row, number(i - 1, j)] = A_w
            else:
                rhs -= A_w * u_exact(0, y)

            # Верхняя точка
            if j + 1 <= N2 - 1:
                M[row, number(i, j + 1)] = A_n
            else:
                rhs -= A_n * u_exact(x, l2)

            # Нижняя точка
            if j - 1 >= 1:
                M[row, number(i, j - 1)] = A_s
            else:
                rhs -= A_s * u_exact(x, 0)

            b[row] = rhs

    return M, b


# -----------------------------
# Метод Якоби
# -----------------------------
def jacobi_method(M, b):
    n = len(b)

    y_old = np.zeros(n)
    y_new = np.zeros(n)

    D = np.diag(M)
    R = M - np.diag(D)

    for iteration in range(1, max_iter + 1):
        y_new = (b - R @ y_old) / D

        residual = np.linalg.norm(M @ y_new - b, ord=np.inf)

        if residual < eps:
            return y_new, iteration, residual

        y_old = y_new.copy()

    return y_new, max_iter, residual


# -----------------------------
# Метод скорейшего спуска
# -----------------------------
def steepest_descent_method(M, b):
    # Умножаем систему M y = b на -1
    A = -M
    rhs = -b

    y = np.zeros(len(rhs))

    for iteration in range(1, max_iter + 1):
        r = rhs - A @ y
        Ar = A @ r

        tau = (r @ r) / (Ar @ r)

        y = y + tau * r

        residual = np.linalg.norm(rhs - A @ y, ord=np.inf)

        if residual < eps:
            return y, iteration, residual

    return y, max_iter, residual


# -----------------------------
# Оценка ошибки
# -----------------------------
def max_error(solution):
    max_err = 0.0

    for j in range(1, N2):
        y = j * h2

        for i in range(1, N1):
            x = i * h1
            ind = number(i, j)

            err = abs(solution[ind] - u_exact(x, y))

            if err > max_err:
                max_err = err

    return max_err


# -----------------------------
# Основная часть программы
# -----------------------------
M, b = build_system()

u_jacobi, it_jacobi, res_jacobi = jacobi_method(M, b)
u_sd, it_sd, res_sd = steepest_descent_method(M, b)

err_jacobi = max_error(u_jacobi)
err_sd = max_error(u_sd)

print("Метод Якоби:")
print("Количество итераций:", it_jacobi)
print("Невязка:", res_jacobi)
print("Максимальная ошибка:", err_jacobi)

print()

print("Метод скорейшего спуска:")
print("Количество итераций:", it_sd)
print("Невязка:", res_sd)
print("Максимальная ошибка:", err_sd)
```

```python
import numpy as np
import math

N = 4
h = 1 / N

u = np.zeros((N + 1, N + 1))

# задаем граничные условия
for j in range(N + 1):
    y = j * h
    u[j, 0] = 40 * y**2       # AB: x = 0
    u[j, N] = 40              # CD: x = 1

for i in range(N + 1):
    x = i * h
    u[0, i] = 40 * math.sin(math.pi * x / 2)   # AD: y = 0
    u[N, i] = 40                               # BC: y = 1

eps = 1e-8
err = 1

while err > eps:
    err = 0

    for j in range(1, N):
        for i in range(1, N):
            old = u[j, i]

            u[j, i] = 0.25 * (
                u[j, i - 1] +
                u[j, i + 1] +
                u[j - 1, i] +
                u[j + 1, i]
            )

            err = max(err, abs(u[j, i] - old))

np.set_printoptions(precision=3, suppress=True)

print("Решение, выведенное сверху вниз по y:")
print(u)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math

# Вариант 9
# u_t = u_xx
# u(x,0) = 1 - x + sin(pi*x)
# u(0,t) = 1, u(1,t) = 0

def exact_solution(x, t):
    return 1 - x + np.exp(-math.pi**2 * t) * np.sin(math.pi * x)

# параметры
N = 20          # число шагов по x
T = 0.5         # конечное время
sigma0 = 0.45   # берем меньше 0.5 для устойчивости

h = 1 / N
tau0 = sigma0 * h**2

M = math.ceil(T / tau0)
tau = T / M

sigma = tau / h**2

print("h =", h)
print("tau =", tau)
print("sigma =", sigma)
print("Количество шагов по времени =", M)

# сетка по x
x = np.linspace(0, 1, N + 1)

# начальное условие
u = 1 - x + np.sin(math.pi * x)

# граничные условия
u[0] = 1
u[N] = 0

# явная схема
for j in range(M):
    old = u.copy()

    for i in range(1, N):
        u[i] = old[i] + sigma * (old[i - 1] - 2 * old[i] + old[i + 1])

    # граничные условия
    u[0] = 1
    u[N] = 0

# точное решение в момент T
u_exact = exact_solution(x, T)

# погрешность
error = np.abs(u - u_exact)

print("Максимальная погрешность =", np.max(error))

print()
print("   x        численное        точное        погрешность")
for i in range(0, N + 1, 5):
    print(f"{x[i]:.2f}    {u[i]:.6f}       {u_exact[i]:.6f}     {error[i]:.6f}")

# график
plt.plot(x, u, 'o-', label='Численное решение')
plt.plot(x, u_exact, '--', label='Точное решение')
plt.xlabel('x')
plt.ylabel('u(x,t)')
plt.title('Лабораторная работа 7, вариант 9')
plt.grid()
plt.legend()
plt.show()
```
