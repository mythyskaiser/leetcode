```python
import numpy as np
import matplotlib.pyplot as plt

L = 1
T = 0.1
N = 50
h = L/N
r = 0.4
tau = r*h**2

M = int(T/tau)
tau = T/M
r = tau/h**2

x = np.linspace(0, L, N+1)
t = np.linspace(0, T, M+1)


def u0(x):
    return 1-x+np.sin(np.pi*x)


def left_bc(t):
    return 1


def right_bc(t):
    return 0


def exact(x, t):
    return 1-x+np.exp(-np.pi**2*t)*np.sin(np.pi*x)


y = np.zeros((M+1, N+1))

for i in range(N+1):
    y[0, i] = u0(x[i])

y[0, 0] = left_bc(0)
y[0, N] = right_bc(0)

for j in range(M):
    y[j+1, 0] = left_bc(t[j+1])
    y[j+1, N] = right_bc(t[j+1])

    for i in range(1, N):
        y[j+1, i] = y[j, i]+r*(y[j, i + 1]-2*y[j, i]+y[j, i-1])


y_true = exact(x, T)

error = np.max(np.abs(y[M]-y_true))

print(error)

plt.plot(x, y_true)
plt.plot(x, y[M], "o")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math

a = 2.0
b = 1.0
T = 1.0

def u(x, y, t):
    return x * math.exp(t) + y**3

def func(x, y, t):
    return x * math.exp(t) - 6 * y

def phi(x, y):
    return u(x, y, 0)

def mu1(y, t):
    return u(0, y, t)

def mu2(y, t):
    return u(a, y, t)

def mu3(x, t):
    return u(x, 0, t)

def mu4(x, t):
    return u(x, b, t)

Nx = 30
Ny = 30

hx = a / Nx
hy = b / Ny

tau0 = 0.5 / (1 / hx**2 + 1 / hy**2)
Nt = int(T / tau0)
tau = T / Nt

x = np.linspace(0, a, Nx + 1)
y = np.linspace(0, b, Ny + 1)

U = np.zeros((Nx + 1, Ny + 1))

for i in range(Nx + 1):
    for j in range(Ny + 1):
        U[i, j] = phi(x[i], y[j])

def borders(U, t):
    for j in range(Ny + 1):
        U[0, j] = mu1(y[j], t)
        U[Nx, j] = mu2(y[j], t)

    for i in range(Nx + 1):
        U[i, 0] = mu3(x[i], t)
        U[i, Ny] = mu4(x[i], t)

borders(U, 0)

for n in range(Nt):
    t = n * tau

    U2 = U.copy()

    for i in range(1, Nx):
        for j in range(1, Ny):

            u_xx = (U[i + 1, j] - 2 * U[i, j] + U[i - 1, j]) / hx**2
            u_yy = (U[i, j + 1] - 2 * U[i, j] + U[i, j - 1]) / hy**2

            U2[i, j] = U[i, j] + tau * (
                u_xx + u_yy + func(x[i], y[j], t)
            )

    borders(U2, (n + 1) * tau)

    U = U2
    if n == 0:
        U_first = U2
    elif n == Nt-1:
        U_last = U2

U_ex = np.zeros((Nx + 1, Ny + 1))

for i in range(Nx + 1):
    for j in range(Ny + 1):
        U_ex[i, j] = u(x[i], y[j], T)

error = np.max(np.abs(U - U_ex))

print(error)

plt.imshow(U, extent=[0, a, 0, b])
plt.show()
```
