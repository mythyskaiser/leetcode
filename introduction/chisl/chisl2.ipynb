## row func

```python
import math
import numpy as np
import matplotlib.pyplot as plt
data = [[1, 4.95], [1.5,3.3], [2,4.05], [2.5,3.85], [3,3.6], [3.5,3.55], [4,3.45], [4.5,3.4]]

def float_range(start,begin,step):
    while start < begin:
        yield start
        start += step

besties = []

def pow_method_lowest_squares(data, num, diviator, sigmoid):
    start = num-100*diviator
    end = num+100*diviator
    arr_errors = {}
    # arr_errors_str = []
    for i in float_range(start, end, diviator):
        error = 0
        for j in range(len(data)):
            func = pow(data[j][0], i)
            error += pow(data[j][1]-func, 2)
        # arr_errors_str.append(f"{error} {i}")
        arr_errors.update({i:error})
    num = min(arr_errors, key=arr_errors.get)
    print(num)
    error = arr_errors[num]
    besties.append(num)
    print(arr_errors)
    
    if(abs(arr_errors[num]-arr_errors[num-diviator])<sigmoid):
        print("done")
        return num, diviator, error , arr_errors
    diviator = diviator/10
    return pow_method_lowest_squares(data, num, diviator, sigmoid)

print(pow_method_lowest_squares(data, 2, 1, 0.00000001))
x = np.linspace(1,5, 1000)
n = 1
for i in range(len(besties)):
    plt.plot(x, pow(x, besties[i]), label = n)
    n+=1
plt.grid(True)
plt.legend()
plt.show()
```
