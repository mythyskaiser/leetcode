```python
import pandas as pd

def selectData(students: pd.DataFrame) -> pd.DataFrame:
    return students[students['student_id']==101]
```

```python
df = pd.DataFrame([1,2,3,4,5,101], columns=['student_id'])
selectData(df)
```
