```python
import random
def sort(a: list):
    k = 0
    for i in range(len(a)):
        min = a[0]
        ind = 0
        for j in range(len(a)-k):
            if a[j] < min:
                ind = j
                min = a[j]
        k += 1
        g = a[ind]
        a[ind] = a[len(a)-k]
        a[len(a)-k] = g
    return a[::-1]
a = [random.randint(1,100) for i in range(30)]
print(sort(a))
```

```python
import random
def binarysearch(a: list, b: int) -> int:
    index_end = len(a)-1
    index_begin = 0
    index_part = int(index_end/2)
    step = 0
    while True:
        if a[index_part] == b:
            return index_part
        elif a[index_end] == b:
            return index_end
        elif a[index_begin] == b:
            return index_begin
        elif a[index_part] < b:
            print("higher", step, index_begin, index_part, index_end)
            index_begin = index_part
            index_part = int((index_end-index_part)/2+index_part)
            index_end = index_end
        elif a[index_part] > b:
            print("lower", step, index_begin, index_part, index_end)
            index_begin = index_begin
            index_end = index_part
            index_part = int(index_part-(index_part-index_begin)/2)
        step += 1
            

a = [random.randint(1,10000000) for i in range(300000)]
a.sort()
b = a[3]
binarysearch(a,b)
```

# 643. Maximum Average Subarray I

```python
class Solution(object):
    def findMaxAverage(self, nums: list[int], k: int) -> float:
        """
        :type nums: List[int]
        :type k: int
        :rtype: float
        """
        max = nums[0]
        n = 1.0
        for i in range(1,k):
            n += 1.0
            max += nums[i]
        avg = max/n
        for i in range(k,len(nums)):
            max += nums[i]
            max -= nums[i-k]
            if (avg < max/n):
                avg = max/n
        return avg

nums = [0,1,1,3,3]
k = 4
Solution.findMaxAverage(1,nums,k)
```

# 3.7 Palyndrome

```python
def palyndrome(s: str) -> bool:
    if s[0] != s[-1]:
        print('False')
        return False
    elif (len(s) == 1) or (len(s)==2 and s[0] == s[-1]):
        print('True')
        return True
    else:
        s = s[1:(len(s)-1)]
        print(s)
        palyndrome(s)
palyndrome("abbccbba")
```

```python
#1
def counter(numbers: list, number: int) -> int:
    k = 0
    for i in range(len(numbers)):
        if number == numbers[i]:
            k += 1
    return k
counter([1,2,3,1,1],1)
```

```python
#2
def second(numbers: list) -> int:
    max_num = max(numbers)
    numbers.remove(max_num)
    second_max = max(numbers)
    return second_max
second([1,1,2,3,99])
```

```python
#3
def slicer(numbers: list) -> list:
    min_num = min(numbers)
    index = 0
    for i in range(len(numbers)):
        if min_num == numbers[i]:
            index = i
    numbers_2 = numbers[index:]
    max_2 = max(numbers_2)
    numbers.remove(max_2)
    return numbers
slicer([1,2,3,3,4,0,9,9,10,1])
```

```python
#4
def righter(numbers: list, k: int) -> list:
    left_num = numbers[-k:]
    right_num = numbers[:-k]
    numbers = left_num + right_num
    return numbers
righter([1,2,3,4,5,6,7,8],4)
```

```python
#5
import numpy as np

def matrix_quad(n: int) -> np.array:
    matrix = np.zeros((n,n))
    for i in range(n):
        for j in range(n):
            if (i==j):
                matrix[i,j] = 1
            elif (i>j):
                matrix[i,j] = 2
            elif (i<j):
                matrix[i,j] = 0
    return matrix.T
matrix_quad(6)
```

```python
#8
def snake_straight(n: int, m: int) -> np.array:
    snake = np.zeros((n,m))
    k = 0
    for i in range(n):
        for j in range(m):
            if(i%2==0):
                snake[i,j] = k
            else:
                snake[i,-j] = k
            k += 1
    return snake
snake_straight(5,5)
```
