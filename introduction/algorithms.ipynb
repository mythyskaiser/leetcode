```python
def getstringnarabu(a: str, p: str):
    result = ""
    if (len(a)>len(p)):
        result = "p"
        p = p.replace('p', '',1)
    while(len(a)>len(p) | len(p) != 0):
        a = a.replace('aa','',1)
        p = p.replace('p', '',1)
        result = result + "aa"
        result = result + "p"

    while(a!="" ):
        if len(p)!=0:
            result = result + 'p'
            p = p.replace('p','',1)
        result = result + 'a'
        a = a.replace('a','',1)
    print(result)
def narabu(n,m):
    ispossible = False
    if n >= m:
        a = "a"*n
        p = "p"*m
    else:
        a = "a"*m
        p = "p"*n
    if int(len(a)/2) <= (len(p)-1):
        ispossible = True
    if ispossible:
        getstringnarabu(a,p)
        return True
    else:
        return False
narabu(5,6)
```

```python
import random
def sort(a: list):
    k = 0
    for i in range(len(a)):
        min = a[0]
        ind = 0
        for j in range(len(a)-k):
            if a[j] < min:
                ind = j
                min = a[j]
        k += 1
        g = a[ind]
        a[ind] = a[len(a)-k]
        a[len(a)-k] = g
    return a[::-1]
a = [random.randint(1,100) for i in range(30)]
print(sort(a))
```

# Бинарный поиск

```python
import random
def binarysearch(a: list, b: int) -> int:
    index_end = len(a)-1
    index_begin = 0
    index_part = int(index_end/2)
    step = 0
    while True:
        if a[index_part] == b:
            return index_part
        elif a[index_end] == b:
            return index_end
        elif a[index_begin] == b:
            return index_begin
        elif a[index_part] < b:
            print("higher", step, index_begin, index_part, index_end)
            index_begin = index_part
            index_part = int((index_end-index_part)/2+index_part)
            index_end = index_end
        elif a[index_part] > b:
            print("lower", step, index_begin, index_part, index_end)
            index_begin = index_begin
            index_end = index_part
            index_part = int(index_part-(index_part-index_begin)/2)
        step += 1
            

a = [random.randint(1,10000000) for i in range(300000)]
a.sort()
b = a[3]
binarysearch(a,b)
```

```python
def binarysearch_upgraded(a: list, b: int) -> int:
    index_begin = 0
    index_end = len(a)-1
    while index_begin<=index_end:
        index_part = int((index_end+index_begin)/2)
        if a[index_part] < b:
            print("higher", index_begin, index_part, index_end)
            index_begin = index_part + 1
        elif a[index_part] > b:
            print("lower", index_begin, index_part, index_end)
            index_end = index_part - 1
        elif a[index_part] == b:
            return index_part

a = [random.randint(1,10000000) for i in range(300000)]
a.sort()
b = a[299999]
binarysearch_upgraded(a,b)
```
